## Balanceamento dos dados
### Alvo: números com muitas casas decimais e variáveis categóricas

### Criação da coluna categórica

```python
import pandas as pd

dataset = pd.read_csv("csv_malware/IIoT_Malware_Timeseries_Dataset.csv")

print(dataset.columns.tolist())
# Contar os totais de cada tipo de Malware no dataset
print(dataset["Label"].value_counts(dropna=False))

# Converter  
dataset["tipo_conexao"] = dataset["Label"].apply(
    lambda valor: "Benign" if valor == "Benign" else "Malign"
)
# Contar os totais de cada tipo de conexão no dataset
print(dataset["tipo_conexao"].value_counts())

# Remover a coluna "Label" do dataset
dataset = dataset.drop(columns=["Label"])
```

### Tratamento dos números com muitas casas decimais

O arredondamento é aplicado apenas às colunas numéricas contínuas (`float`). Colunas inteiras, como portas e horários, e colunas categóricas não são alteradas. Neste projeto serão mantidas quatro casas decimais.

```python
# Identificar automaticamente as colunas que possuem números decimais
colunas_decimais = dataset.select_dtypes(include=["float"]).columns

print("Colunas tratadas:")
print(colunas_decimais.tolist())

# Manter quatro casas decimais
dataset[colunas_decimais] = dataset[colunas_decimais].round(4)

# Conferir uma amostra após o tratamento
display(dataset[colunas_decimais].head())

# Salvar somente depois que todos os tratamentos forem aplicados
caminho_saida = "csv_malware/IIoT_Malware_Timeseries_Dataset_modificado.csv"
dataset.to_csv(caminho_saida, index=False)
print(f"Dataset tratado salvo em: {caminho_saida}")
```
